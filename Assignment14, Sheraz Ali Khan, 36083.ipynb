```python
# Question 1: Which is the most popular product category ?
# Answer: Home Doctor
df.groupby("product_category").count()
```

```python
# Question 2: What is the average order value?
# Answer: Karachi
df.groupby("city").count()
```

```python
# Question 3: Most used payment method:
# Cash on delivery
df.groupby("payment_method").count()
```

```python
# Bar chart
avg_city = df.groupby('city')['order_value'].mean()
```

```python
import matplotlib.pyplot as plt
categories = avg_city.index
values = avg_city.values
plt.bar(categories, values, color = 'red', width = 0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.show()
```
