```python
# Histogram
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))

sns.histplot(data = df, x='daily_transport_cost',bins = 10, kde = True)

plt.title('Distribution of daily transpot cost')
plt.xlabel('Daily Transport cost')
plt.ylabel('Freaquency')
plt.show()
```

```python
#Question5:Compare average internet usage by gender.
#Answer:
gender_internet = df.groupby('gender')['internet_usage'].mean()

gender_internet.plot(kind='bar', figsize=(8,5))

plt.title('Average Internet Usage by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Internet Usage (Hours)')
plt.xticks(rotation=0)

plt.show()
```
